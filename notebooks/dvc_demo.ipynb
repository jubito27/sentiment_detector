```python
import pandas as pd
df = pd.read_csv('/mnt/WindowsD/Artificial Intelligence/Datasets/tweets_emotions.csv')

df.drop(columns=['tweet_id' ,'Unnamed: 0' ] , inplace=True)
```

```python
df.head()
```

```python
df = df[(df['sentiment'] == 'sadness') | (df['sentiment'] == 'happiness') | (df['sentiment']  == 'neutral') | (df['sentiment']  == 'anger')]
```

```python
df.shape
```

```python
df = df.sample(3000)
```

```python
x_ = df['content']
y_ = df['sentiment']
```

```python
y_.unique()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(y_)
y
```

```python
from automac.text import transform

x = x_.apply(transform)
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
x_train_ , x_test_ , y_train_, y_test_ = train_test_split(x , y , test_size=0.2 , random_state=42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tf = TfidfVectorizer()
x_trans = tf.fit_transform(x_train_)
```

```python
pd.DataFrame(x_trans.toarray())
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()
x_train = cv.fit_transform(x_train_)
x_test = cv.transform(x_test_)
```

```python
train = pd.DataFrame(x_train.toarray())
```

```python
test = pd.DataFrame(x_test.toarray() )
```

```python
import xgboost

model = xgboost.XGBClassifier()
model.fit(train , y_train_)
```

```python
import lightgbm

model2 = lightgbm.LGBMClassifier()
model2.fit(train , y_train_)
```

```python
from sklearn.metrics import accuracy_score , classification_report , precision_score , recall_score , f1_score , auc , roc_auc_score

y_pred = model.predict(test)
print(accuracy_score(y_test_ , y_pred))
print(classification_report(y_test_ , y_pred))
```

```python
y_pred = model2.predict(test)
print(accuracy_score(y_test_ , y_pred))
print(classification_report(y_test_ , y_pred))
```

```python


print("Precision : " , precision_score(y_test_ , y_pred , average='macro'))
print("Recall : " , recall_score(y_test_ , y_pred, average='macro'))
print("f1_score : " , f1_score(y_test_ , y_pred, average='macro'))
y_proba = model2.predict_proba(test)  # shape: (n_samples, n_classes)

print("AUC : ", roc_auc_score(y_test_, y_proba, average='macro', multi_class='ovr'))
```

```python
import pickle as pk
import numpy as np
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
)

model_path = "/mnt/WindowsD/Artificial Intelligence/MLOPS/DVC/dvc_demo/model/model.pkl"

with open(model_path, "rb") as f:
    model = pk.load(f)

x_test = pd.read_csv(
    "/mnt/WindowsD/Artificial Intelligence/MLOPS/DVC/dvc_demo/data/transformed/x_test.csv",
    index_col=0,
)
y_test = pd.read_csv(
    "/mnt/WindowsD/Artificial Intelligence/MLOPS/DVC/dvc_demo/data/transformed/y_test.csv",
    index_col=0,
).squeeze("columns")

print(x_test.shape, y_test.shape)

y_pred = model.predict(x_test)
ac = accuracy_score(y_test, y_pred)
cr = classification_report(y_test, y_pred)
precision = precision_score(y_test, y_pred, average="macro")
recall = recall_score(y_test, y_pred, average="macro")
f1 = f1_score(y_test, y_pred, average="macro")

y_proba = model.predict_proba(x_test)
model_classes = np.asarray(model.classes_)
test_classes = pd.unique(y_test)

if len(test_classes) < 2:
    raise ValueError("ROC AUC requires at least two classes in y_test.")

if not np.isin(test_classes, model_classes).all():
    raise ValueError("y_test contains a class that the model was not trained to predict.")

# Compute one-vs-rest AUC only for classes represented in y_test.
class_aucs = []
for cls in test_classes:
    class_index = np.flatnonzero(model_classes == cls)[0]
    y_binary = (y_test == cls).astype(int)
    class_aucs.append(roc_auc_score(y_binary, y_proba[:, class_index]))

auc = float(np.mean(class_aucs))

data = {
    "accuracy": ac,
    "precision": precision,
    "recall": recall,
    "f1": f1,
    "auc": auc,
}

df = pd.DataFrame([data] , index=['lighgbm'])
print(cr)
print(df)
```
